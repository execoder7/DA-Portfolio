### 1. SLA Compliance by Vendor

```python
sla_by_vendor = df.groupby("Vendor")['SLA_Met'].mean().reset_index()
```

```python
fig1 = px.bar(sla_by_vendor, x='Vendor',y='SLA_Met',title="SLA Compliance by Vendor")
fig1.show()
```

### 2. Attrition Risk by Role

```python
attrition_by_role = df.groupby(['Role','Attrition_Risk']).size().reset_index(name="Count")
```

```python
df.groupby('Role')['Attrition_Risk'].size().reset_index()
```

```python
fig2 = px.bar(attrition_by_role, x='Role', y='Count', color='Attrition_Risk', barmode='stack', title="Attrition Risk by Role")
fig2.show()
```

### 3. Performance vs Saudization

```python
performance_by_nationalization = df.groupby('Saudization_Status')['Performance_Score'].mean().reset_index()
```

```python
fig3 = px.bar(performance_by_nationalization, x='Saudization_Status', y='Performance_Score', title='Performance Score by Saudization')
fig3.show()
```

### 4. Cost per Perfomance Point (Value for Money)

```python
df['Cost_Performance_Ratio'] = df['Hourly_Rate_SAR'] / df['Performance_Score']
```

```python
cost_efficiency_by_vendor = df.groupby("Vendor")['Cost_Performance_Ratio'].mean().reset_index()
```

```python
fig4 = px.bar(cost_efficiency_by_vendor, x='Vendor', y='Cost_Performance_Ratio', title='Cost per Performance')
fig4.show()
```
